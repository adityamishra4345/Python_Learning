```python
import numpy as np

# Data structure: [Restaurant ID, Sales 2020, Sales 2021, Sales 2022, Sales 2023]
restaurant_data = np.array([
    [1, 150000, 120000, 180000, 210000],  # Restaurant 1
    [2, 280000, 200000, 250000, 310000],  # Restaurant 2
    [3, 95000,  85000,  110000, 130000],  # Restaurant 3
    [4, 420000, 390000, 440000, 490000],  # Restaurant 4
    [5, 185000, 160000, 195000, 225000]   # Restaurant 5
])

print("===Restorent Sales Report===")
print("sales data shape",restaurant_data.shape)
print("sales data of first 3 resto. : \n",restaurant_data[:3])#used slicing
print("Restorent data  : \n",restaurant_data[:,1:])#used slicing [rows , column ] 
# here in row : means every thing i.e no start and end point and 
#in column start from 1 and go to end 1:  thus retorent id will be skipped
```

### In NumPy, axis=0 means down the columns (vertical), and axis=1 means across the rows (horizontal).
# normally hum top to bottom jate hai so axes is 0 and if left to right jainge toh axis changed to 1

```python
# total sales  restorent wise
yearly_total=np.sum(restaurant_data[ : ,1: ],axis=1)# axis=0  => addn along rows having all rows and columns from 1 to last
print(yearly_total)
#to get sums of all restorent in year wise
print("year wise collection total regardless of restorent: ",np.sum(restaurant_data[:,1:],axis=0))
```

# minimum sales per restorent

```python
min_sales=np.min(restaurant_data[:,1:],axis=1)
print("Minimum sales per restorent is ",min_sales)
# max sales per year
max=np.max(restaurant_data[:,1:],axis=0)
print("Maximum sales per year is ",max)
```

# Addition and Multiplication

```python
vec1=np.array([1,2,3,4,5])
vec2=np.array([6,7,8,9,10])
print("vector addn: ",vec1 + vec2)
print("vector multiplication: ",vec1 * vec2) # gives multiplication index wise
print("Dot product : ",np.dot(vec1,vec2)) #gives dot product of the both  arrays
angle=np.arccos(np.dot(vec1,vec2)/(np.linalg.norm(vec1)*np.linalg.norm(vec2)))
# formula is used her that angle is (A.B)/(||A||*||B||) and np.arccos converting this cos theta to theta
print("Angle between both vectors in radians : ",angle)
```

```python
food=np.array(['biryani','salsa','samosa','kachori'])
#there are many methods if it is python like str.upper but since its an vector so we have to 
# vectorize this method to use it in the vector
upper=np.vectorize(str.upper)
capital=np.vectorize(str.capitalize)
print("Upper Dishes : ",upper(food))
print("capital Dishes : ",capital(food))
```

Array reshaping


```python
arr=np.arange(12)
print("Array ",arr)
resh=arr.reshape(3,4)
print("Reshaped array  \n",resh)
flat=resh.flatten()
print("flattened ",flat)
raval=resh.ravel()
print("ravel ",raval)
#difference between flattened and ravel is flattened make a flat copy and if anything modified in flat 
# will not reflect in original on other hand ravel just view the orginal in flat form and
# any change will be reflected in the original
  
```

```python
transp = resh.T
print("transposed array  \n",transp)
```

```python
import matplotlib.pyplot as plt
x=['mon','tues','wed','thur','fri','sat','sun']
y=[10,15,7,20,22,1,13]

plt.plot(x,y, color='cyan', linestyle='dashed', marker='')

plt.title('Bakery Sales')

plt.xlabel('Days of the week')
plt.ylabel('sales per day')
```

```python
import matplotlib.pyplot as plt
months=[1,2,3,4]
sales=[10,15,7,8]

plt.plot(months,sales, color='red', linestyle='dotted', marker='o',label='2026 sales data')

plt.title('monthly sales')

plt.xlabel('month no')
plt.ylabel('sales ')
#plt.legend() will do on top right

plt.legend(loc='lower left') 

plt.grid(color='cyan',linestyle='--' ,linewidth=1)

#to lint x and y axis lim used
plt.xlim(1,9)
plt.ylim(1,19)

#alisa axes to namess
plt.xticks([1,2,3,4],['m1',"m2","m3",'m4'])
```
